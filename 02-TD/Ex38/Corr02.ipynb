```python
DrH = (DsubG + Ip + DhydG)
print(DrH)

P = np.poly1d(np.polyfit(DrH, Ec, deg =1))
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
ax.plot(DrH, Ec,  marker = 'o', ls = 'none', color = 'black')
ax.plot(DrH, P(DrH), color = 'red')

ax.set_xlabel("$\Delta_r H^\circ$ [$kJ \cdot mol^{-1}$]")
ax.set_ylabel("$E^\circ$ [$V/ESH$]")

fig.savefig("Corr02.pdf", format = 'pdf')
```
