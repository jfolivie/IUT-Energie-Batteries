# Graph

```python
import numpy as np

T, LiBi, Li3Bi = np.loadtxt("Graph.csv", delimiter = ",", skiprows=1).T
mask_LiBi =np.invert(np.ma.masked_invalid(LiBi).mask)
mask_Li3Bi =np.invert(np.ma.masked_invalid(Li3Bi).mask)

P_LiBi = np.poly1d(np.polyfit(T[mask_LiBi], LiBi[mask_LiBi], deg = 1))
P_Li3Bi = np.poly1d(np.polyfit(T[mask_Li3Bi], Li3Bi[mask_Li3Bi], deg = 1))
```

```python
print(P_LiBi)
print(P_Li3Bi)
```

```python
print(P_LiBi(380+278), P_Li3Bi(380+278))
```
