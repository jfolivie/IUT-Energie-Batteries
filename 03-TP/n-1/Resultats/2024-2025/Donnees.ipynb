```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
groups=("Elvine-Coralie", "Etienne", "Margo-Noemie", "Mehdi-Jade-Lucas-Thashana", 'Irene-Lucile',
       'Aissatou-Clemence', "Mathis-Killian", "Noor-Liticia", "Camille-Adame","YvannaRose-Mouhamed")

datas = [pd.read_excel("Bilan.ods", sheet_name=grp) for grp in groups] 

Ug = [datas[ng]["Up"] + datas[ng]["Uj"] + datas[ng]["Um"] for ng in range(len(datas))]
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.subplots()
for ng in range(len(groups)) :
    ax.plot(datas[ng]['I'], Ug[ng], label = groups[ng])

ax.plot(datas[ng]['I'], 1.1 - 0.520 * datas[ng]['I'], color = 'black')
ax.plot(datas[ng]['I'], 1.1 - 0.26 * datas[ng]['I'], color = 'black')
ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))

ax.xaxis.set_label_coords(1, 0.5)
ax.yaxis.set_label_coords(0.5, 1.0)
ax.set_xlabel("$I ~ [mA]$")
ax.set_ylabel("$U ~ [V]$")
fig.legend()
```
