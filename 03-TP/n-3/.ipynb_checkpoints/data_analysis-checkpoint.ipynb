```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib as mpl
import matplotlib.colors as colors
from matplotlib import rc

##Initiate
rc('text', usetex = True)
rc('font', family='serif', size = 18)
rc('lines', linewidth = 2)
rc('figure', dpi= 100)
rc('axes', lw = 3)
rc(('xtick', 'ytick'), labelsize = 18)
rc('grid', lw = 5)
mpl.interactive(True)

_dpi = 50
```

```python
_t_arg = "Time (s)"
_U_arg = "WE(1).Potential (V)"
path = "./datas/"


I = 0.001 #A

seq_C = ("C1", "C2", "C3", "C4", "C5")
seq_D = ("D1", "D2", "D3", "D4", "D5")

datas_C = [pd.read_excel(path + C + ".ods") for C in seq_C]
datas_C = [data[[_t_arg, _U_arg]] for data in datas_C]

datas_D = [pd.read_excel(path + D + ".ods") for D in seq_D]
datas_D = [data[[_t_arg, _U_arg]] for data in datas_D]
```

# Profil galvanostatique - charge/décharge

```python
I = 0.001 #A

def colormap(num_colors, name = 'jet') : #'Spectral') :
    cmap = mpl.cm.get_cmap(name = name, lut = num_colors)
    return cmap(range(num_colors))

pen_matplotlib = colormap(len(seq_C), name = "copper")

#Conversion temps-courant
def get_Q(t : 's') -> 'mAh': 
    Q = I * t / 3.6
    return Q - Q[0] 

Q_C = [get_Q(data[_t_arg]) for data in datas_C]
Q_D = [get_Q(data[_t_arg]) for data in datas_D]
```

```python
# t - U
fig = plt.figure(figsize=(6,4))
ax=fig.subplots()
for n in range(len(seq_C)) :
    plt1 = ax.plot(datas_C[n][_t_arg],datas_C[n][_U_arg], color = 'red')
    plt2 = ax.plot(datas_D[n][_t_arg],datas_D[n][_U_arg], color = 'blue')

    if n == 0 : 
        plt1[0].set_label("Charge")
        plt2[0].set_label("Décharge")
    
ax.set_xlabel("Temps d'acquisition [$s$]")
ax.set_ylabel("Tension [$V$]")

fig.legend()
fig.tight_layout()

fig.savefig(path + "U-t.pgf", dpi = _dpi)
fig.savefig(path + "U-t.pdf", dpi = _dpi)
```

```python
fig = plt.figure(figsize=(6,4))
ax=fig.subplots()
for n in range(len(seq_C)) :
    ax.plot(Q_C[n],datas_C[n][_U_arg], color = pen_matplotlib[n])
    ax.plot(Q_D[n],datas_D[n][_U_arg], color = pen_matplotlib[n])

    
cax = fig.add_axes([0.84, 0.5, 0.02, 0.25])
pen_matplotlib = list(pen_matplotlib)
cmap = colors.ListedColormap(pen_matplotlib)
norm = colors.BoundaryNorm(np.arange(len(seq_C)+1)+0.5,len(seq_C))
sm = plt.cm.ScalarMappable(cmap = cmap, norm = norm)
sm.set_array([])

cb = fig.colorbar(sm, ticks = np.arange(len(seq_C)+1), cax = cax)
cb.set_label('Numéro du cycle')
#cb.set_ticklabels(self.params)

ax.set_xlabel("Capacité / Charge [$mAh$]")
ax.set_ylabel("Tension [$V$]")

fig.tight_layout()

fig.savefig(path + "U-Q.pgf", dpi = _dpi)
fig.savefig(path + "U-Q.pdf", dpi = _dpi)
```

```python
Qmax_C = np.array([Q_C[n].max() for n in range(len(Q_C))])
Qmax_D = np.array([Q_D[n].max() for n in range(len(Q_D))])

SoC_C = [(Q_C[n] / Q_C[n].max())*100 for n in range(len(Q_C))]
SoC_D = [(1-Q_D[n] / Q_D[n].max())*100 for n in range(len(Q_D))]


fig = plt.figure(figsize=(6,4))
ax=fig.subplots()
for n in range(len(seq_C)) :
    ax.plot(SoC_C[n],datas_C[n][_U_arg], color = pen_matplotlib[n])
    ax.plot(SoC_D[n],datas_D[n][_U_arg], color = pen_matplotlib[n])

    
cax = fig.add_axes([0.84, 0.5, 0.02, 0.25])
pen_matplotlib = list(pen_matplotlib)
cmap = colors.ListedColormap(pen_matplotlib)
norm = colors.BoundaryNorm(np.arange(len(seq_C)+1)+0.5,len(seq_C))
sm = plt.cm.ScalarMappable(cmap = cmap, norm = norm)
sm.set_array([])

cb = fig.colorbar(sm, ticks = np.arange(len(seq_C)+1), cax = cax)
cb.set_label('Numéro du cycle')
#cb.set_ticklabels(self.params)

ax.set_xlabel("State-Of-Charge [$\%$]")
ax.set_ylabel("Tension [$V$]")

fig.tight_layout()

fig.savefig(path + "U-SoC.pgf", dpi = _dpi)
fig.savefig(path + "U-SoC.pdf", dpi = _dpi)
```
